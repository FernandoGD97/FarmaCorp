```python
import os, subprocess
from bs4 import BeautifulSoup
```

```python
BASE_URL = 'https://evademecum.boiron.es'
OUTPUT_FOLDER = 'corpus/vademecum_boiron/'
INFO_FILENAME = 'vad_boiron_info.txt'
SCRIPT_FILENAME = 'vad_boiron.sh'
LINKS_FILENAME = 'vademecum_boiron.html'
```

```python
with open(LINKS_FILENAME) as tf:
    n = 0
    med_page = tf.read()
    med_soup = BeautifulSoup(med_page, 'html.parser')
    med_list = med_soup.findAll("a", class_="medAdj")
    file_info = open(INFO_FILENAME, "w")
    file_script = open(SCRIPT_FILENAME, "w")
    for med in med_list:
        cod = f'VAB{n:05}'
        url = BASE_URL + med["href"]
        file_info.write(cod + "\t" + med["title"] + "\t" + url + "\t" + os.path.basename(med["href"])  + "\n")
        file_script.write("wget " + url + "\n" + "sleep 1" + "\n")
        n += 1
    file_info.close()
    file_script.close()
    print("FIN")
        
```

```python
with open(INFO_FILENAME) as tf:
    for line in tf:
        data = line.split("\t") 
        command = f'mv "{OUTPUT_FOLDER}{data[3].strip()}" "{OUTPUT_FOLDER}{data[0]}.pdf"'
        subprocess.call(command, shell=True)
```

```python
n = 0
for cont in range(9880):
    code = f'VAB{cont:05}'
    if not os.path.exists(OUTPUT_FOLDER + code + '.pdf'):
        print(code)
        n += 1
print("Total:", n)
```

```python
with open(INFO_FILENAME) as tf:
    meta = open("vad_boiron_info_def.txt", "w")
    for line in tf:
        data = line.split("\t") 
        if os.path.exists(OUTPUT_FOLDER + data[0] + '.pdf'):
            meta.write(line)
    meta.close()
```
