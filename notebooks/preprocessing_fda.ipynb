```python
import os, subprocess, time
from bs4 import BeautifulSoup
from user_agents import get_random_user_agent
```

```python
BASE_URL = 'https://dps.fda.gov/'
OUTPUT_FOLDER = 'corpus/fda/'
INFO_FILENAME = 'fda_info.txt'
LINKS_FILENAME = 'fda_links.txt'
SCRIPT_FILENAME = 'fda.sh'
```

```python
with open(LINKS_FILENAME) as tf:
    for url in tf:
        random_user_agent = get_random_user_agent()
        command = f'wget --user-agent="{random_user_agent}" --directory-prefix={OUTPUT_FOLDER} {url}'
        subprocess.call(command, shell=True)
        time.sleep(1)
```

```python
with open(INFO_FILENAME) as tf:
    for line in tf:
        data = line.split("\t") 
        command = f'mv "{OUTPUT_FOLDER}{data[3].strip()}" "{OUTPUT_FOLDER}{data[0]}.pdf"'
        subprocess.call(command, shell=True)
```

```python
n = 0
for cont in range(961):
    code = f'FDA{cont:05}'
    if not os.path.exists(OUTPUT_FOLDER + code + '.pdf'):
        print(code)
        n += 1
print("Total:", n)
```

```python
with open(INFO_FILENAME) as tf:
    meta = open("fda_info_def.txt", "w")
    for line in tf:
        data = line.split("\t") 
        if os.path.exists(OUTPUT_FOLDER + data[0] + '.pdf'):
            meta.write(line)
    meta.close()
```
